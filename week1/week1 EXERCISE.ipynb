To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
# setup

load_dotenv(override=True)

GPT_MODEL = 'deepseek-chat'
GPT_URL = 'https://api.deepseek.com'
OLLAMA_MODEL = 'llama3.2'
OLLAMA_URL = "http://localhost:11434/v1"
```

```python
# set up environment
def get_openapi_client(model):
    if model == GPT_MODEL:
        return OpenAI(base_url=GPT_URL)
    return OpenAI(base_url=OLLAMA_URL)
```

```python
# here is the question; type over this to ask something new

context = """
I want you to act as a expert software engineer to explain some code to me, a rookie software developer.
Please make the explanation simple and provide examples when i need it.

Important: You need to only answer software engineering questions.
"""

def explain_code_question(model, question):
    openai = get_openapi_client(model)
    stream = openai.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": context},
            {"role": "user", "content": question}
      ],
        stream=True
    )

    for chunk in stream:
        print(chunk.choices[0].delta.content, end="")

```

```python
# Get gpt-4o-mini to answer, with streaming
question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""

explain_code_question(GPT_MODEL, question)
```

```python
# Get Llama 3.2 to answer

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""

explain_code_question(OLLAMA_MODEL, question)
```
